```python
X1, Y1 = 2, 1
X2, Y2 = 5, 5
A, B, C, D = 0, 5, 1, 5
def f1(x, y):
    global X1, Y1
    return (x - X1)**2 + (y - Y1)**2

def f2(x, y):
    global X2, Y2
    return (x - X2)**2 + (y - Y2)**2
```

```python
X = [x for x in range(A, B+1)]
Y = [y for y in range(C, D+1)]

import matplotlib.pyplot as plt
import numpy as np

def line(x1, y1, x2, y2):
    x_for_line = np.linspace(min(x1, x2) , max(x1, x2), 100)
    k = (y2 - y1) / (x2 - x1)
    b = y1 - k * x1 
    y_for_line =  k*x_for_line + b
    plt.plot(x_for_line, y_for_line, color="black")
    

line(X1, Y1, X2, Y2)
for x in X:
    for y in Y:
        plt.scatter(x, y, c="black")
plt.grid()
plt.show()

```

```python
fs = []
len_cell = 4
print("|" + "x" + " "*((len_cell-1)) +"|"+ "y" + " "*(len_cell-1) + "|"+ "f1" + " "*(len_cell-2) + "|"+ "f2" + " "*(len_cell-2)  + "|"+ "num" )
i = 1
for x in X:
    for y in Y:
        print("|" + str(x), end=" "*(len_cell-len(str(x))))
        print("|" + str(y), end=" "*(len_cell-len(str(y))))
        f1_temp = f1(x,y)
        f2_temp = f2(x,y)
        print("|" + str(f1_temp), end=" "*(len_cell-len(str(f1_temp))))
        print("|" + str(f2_temp), end=" "*(len_cell-len(str(f2_temp))))
        print("|" + str(i) , end='\n')
        fs.append([f1_temp, f2_temp])
        plt.scatter(f1_temp, f2_temp, c="black")
        plt.text(f1_temp+0.3, f2_temp+0.3, str(i), 
             ha='center', va='bottom',  
             fontsize=8, color='red')
        i += 1
    

pareto = []
for idx, pt in enumerate(fs):
    dominated = False
    for jdx, other in enumerate(fs):
        if idx != jdx and other[0] <= pt[0] and other[1] <= pt[1] and (other[0] < pt[0] or other[1] < pt[1]):
            dominated = True
            break
    if not dominated:
        pareto.append(pt)
        

pareto = sorted(pareto, key=lambda p: p[0])


for pt in pareto:
    plt.scatter(pt[0], pt[1], c="green")
for i in range(len(pareto)-1):
    line(pareto[i][0], pareto[i][1], pareto[i+1][0], pareto[i+1][1])
plt.grid()
plt.show()
```
